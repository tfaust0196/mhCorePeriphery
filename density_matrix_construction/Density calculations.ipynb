```python
import numpy as np
from scipy import stats
numNodes = 76
numLayers = 5
densities = []
groups = np.loadtxt('lukegroupassigns.txt', dtype=int)
for l in range(numLayers):
    adj_matrix = np.loadtxt('newlukelayer' + str(l+1) + '.txt', dtype=int)
    density = np.zeros((8,8))
    for g1 in range(8):
        indices1 = np.where(groups[:,l] == g1)[0]
        for g2 in range(8):
            indices1 = np.where(groups[:,l] == g1)[0]
            indices2 = np.where(groups[:,l] == g2)[0]
            edges = 0
            for i in indices1:
                for j in indices2:
                    if adj_matrix[i,j] >= 0.5:
                        edges = edges + 1
            if g1 == g2:
                if len(indices1) <= 1:
                    density[g1,g2] = -1
                else:
                    density[g1,g2] = edges/(len(indices1) * (len(indices1) - 1))
            else:
                if len(indices1) == 0 or len(indices2) == 0:
                    density[g1,g2] = -1
                else:
                    density[g1,g2] = edges/(len(indices1) * len(indices2))
    densities.append(density)
for l in range(numLayers):
    np.savetxt('lukedensitieslayer' + str(l+1) + '.txt', densities[l])
```
